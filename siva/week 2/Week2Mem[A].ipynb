```python
import pandas as pd
```

```python
# ##Step 1: Loading the Dataset

# ##Purpose:
# The cleaned retail dataset is loaded into Python using Pandas to perform cohort analysis.
```

```python
df=pd.read_csv(r"D:\Datasets\cleaned_retail_data.csv")
```

```python
df.head()
```

```python
df.columns
```

```python
df[['TransactionMonth','CohortMonth']].head()
```

```python
# Step 2: Converting Month Columns to Date Format

# Purpose:
# TransactionMonth and CohortMonth are converted into datetime format so that month-based calculations can be performed accurately.
```

```python
df['TransactionMonth']=pd.to_datetime(df['TransactionMonth'])
df['CohortMonth']=pd.to_datetime(df['CohortMonth'])
```

```python
# Step 3: Calculating Cohort Index

# Purpose:
# CohortIndex represents the number of months elapsed since a customer's first purchase.
```

```python
df['CohortIndex'] = (
    (df['TransactionMonth'].dt.year - df['CohortMonth'].dt.year) * 12
    + (df['TransactionMonth'].dt.month - df['CohortMonth'].dt.month)
)
```

```python
# Step 4: Verifying Cohort Index

# Purpose:
# To ensure that the CohortIndex has been calculated correctly for each transaction.
```

```python
df[['TransactionMonth','CohortMonth','CohortIndex']].head()
```

```python
# Step 5: Grouping Customers

# Purpose:
# Customers are grouped according to their CohortMonth and CohortIndex.
```

```python
cohort_data = df.groupby(
    ['CohortMonth', 'CohortIndex']
)['CustomerID'].nunique().reset_index()
```

```python
# Step 8: Final Cohort Table

# Purpose:
# The final cohort table is generated and used for retention analysis and cohort matrix creation.
```
