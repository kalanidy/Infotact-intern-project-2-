# Cohort Grouping (groupby)
In this section, we load the cleaned dataset fro week 1 and use Pandas groupby to group customers by CohortMonth and CohortIndex. CohortIndex represents the number of months since a customer's first purchase. Ths forms the foundation of the retention matrix.

```python
import pandas as pd
df = pd.read_csv("cleaned_retail_data_updated.csv", encoding='latin-1')
print(df.head())
print(df.shape)
```

```python
cohort_data = (
    df.groupby(['CohortMonth', 'CohortIndex'])['CustomerID']
      .nunique()
      .reset_index()
)
```

# Pivot Table & Retention Matrix
We used Pandas pivot_table to reshape the grouped cohort data into a structured retention matrix. Columns were renamed to Month 0, Month 1, Month 2 etc. for readability. We also validated that Month 0 always equals the full original cohort size.

```python
retention_matrix = cohort_data.pivot_table(
    index='CohortMonth',
    columns='CohortIndex',
    values='CustomerID'
)
```

```python
retention_matrix.columns = [
    f"Month {int(col)}" for col in retention_matrix.columns
]
```

```python
print("Original Cohort Sizes:")
print(retention_matrix["Month 0"])
if retention_matrix["Month 0"].isnull().sum() == 0:
    print("\n✓ Validation Passed: Month 0 contains the full cohort size for every cohort.")
else:
    print("\n✗ Validation Failed: Some cohorts have missing Month 0 values.")
```

```python
print("\nRetention Matrix:")
print(retention_matrix)
```

```python
retention_matrix.to_csv("retention_matrix.csv")
```

```python
print("\nRetention matrix saved as retention_matrix.csv")
```

##  Absolute Retained Users
In this section we examine the absolute number of customers retained in each cohort month. Month 0 represents the original cohort size when the customers first made a purchase. We also validate the retention Matrix to ensure no month exceeds the orginal cohort size (Month 0), and sumamarize the total retained users across all cohorts per month

```python
print("Cohort Retention Matrix - Absolute Users")
print(retention_matrix)
```

```python
print("Month 0 - Original Cohort Size:")
print(retention_matrix["Month 0"])
```

```python
print("Total Retained Users Per Month:")
print(retention_matrix.sum())
```

```python
# Validate Matrix
valid = (retention_matrix.drop(columns=["Month 0"])
         .le(retention_matrix["Month 0"], axis=0)
         .all().all())
print("Matrix Valid (no month exceeds Month 0):", valid)
print("Matrix shape:", retention_matrix.shape)
```

```python
# Summary statistics
print("\nRetained Users Summmary:")
print(retention_matrix.describe().round(2))
```

```python
# Save retained users summary to csv
summary = retention_matrix.describe().round(2)
summary.to_csv("retained_users_summary.csv")

print("Summary saved as retained_users_summary.csv")
```

```python
# save total retained user per month to csv
total_per_month = retention_matrix.sum().reset_index()
total_per_month.columns =["Month", "Total Retained Users"]
total_per_month.to_csv("total_retained_per_month.csv", index=False)
print("Total retained per month saved as total_per_month.to_csv")
```

# Percentage Rentention Rate & Documentation
In this section, we calculate the percentage retention rate for each cohort by dividing each month's retained users by the original cohort size (Month 0). This tells us what percentage of customers came back each month after their first purchase.

```python
# Calculate Retenteion Percentage Rate
retention_rate = retention_matrix.divide(
    retention_matrix["Month 0"], axis=0
).round(4)*100

print("Percentage Retention Rate Matrix:")
print(retention_rate.round(2))
```

```python
# Validate Percentage values
print("Month 0 Retention (should all be 100%):")
print(retention_rate["Month 0"])

valid = (retention_rate.fillna(0) <= 100).all().all()
print("\nAll values within 0-100%:", valid)
```

```python
# Identify Highest churn month
avg_retention = retention_rate.mean()
churn = avg_retention.diff().dropna()
highest_churn_month = churn.idxmin()

print("Month with highest Churn:", highest_churn_month)
print("\nAverage Retention Rate per month:")
print(avg_retention.round(2))
```

```python
retention_rate.to_csv("percentage_retention_rate.csv")
print("Percentage retention rate saved as percentage_retention_rate.csv")
```

## Kalanidy: Percentage Retention Rate & Documentation

### What was done:
We calculated the percentage retention rate by dividing each 
cohort month's absolute users by Month 0 value and multiplying 
by 100. Month 0 was confirmed as 100% for all cohorts. NaN values 
were filled with 0 before validation to handle cohorts with 
incomplete months. The month with the highesr churn was identified 
by finding the steepest average retention drop. results were 
saved as CSV.

### Why it was done:
Percentage retention rates are more meaningful than absolute 
numbers because they allow fair comparison between cohorts 
of different sizes. A cohort of 500 customers retaining 100 
is very different from a cohort of 200 retaining 100 — 
percentages reveal the true retention story. Identifying the
highest churn month gives the Product Manager actionable insight
on when to intervene.

### Steps:
1. Divide each row of retention matrix by its Month 0 value
2. Multiply by 100 to convert to percentage
3. Validate Month 0 equals 100% for all cohorts
4. Confirm all values fall within 0-100% range
5. Calculate average retention rate per month across all cohorts
6. Identify the month with the steepest drop (highest churn)
7. Save percentage retention matrix as percentage_retention_rate.csv

### Note on NaN Values
NaN values in the matrix are expected and not errors. They appear
in later months for recent cohorts because those months have not
yet occured in the dataset timeline.

### Output:
- Percentage retention rate matrix
- Highest churn month identified
- percentage_retention_rate.csv saved
