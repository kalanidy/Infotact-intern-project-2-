```python
import pandas as pd
df = pd.read_csv("C:/Users/yash6/Internship 1/OnlineRetail.csv", encoding='latin-1')
print(df.head())
print(df.shape)
```

```python
# Correct syntax - notice the closing bracket ]
df = df[~df['InvoiceNo'].astype(str).str.startswith('C')]
# Alternative ways to write the same filtering:
# Method 1: More readable multi-line approach
mask = df['InvoiceNo'].astype(str).str.startswith('C')
df = df[~mask]
# Method 2: Using query method
df = df.query("not InvoiceNo.str.startswith('C')", engine='python')
# Method 3: Using loc for explicit filtering
df = df.loc[~df['InvoiceNo'].astype(str).str.startswith('C')]
# Verify the filtering worked
print("Dataset shape after removing cancelled transactions:", df.shape)
print("Sample InvoiceNo values:", df['InvoiceNo'].head().tolist())
```

```python
import pandas as pd

# Assuming your dataset is in a DataFrame called 'df'
# Remove rows where CustomerID is null/NaN
df_cleaned = df.dropna(subset=['CustomerID'])

# Alternative method - remove rows where CustomerID is null
df_cleaned = df[df['CustomerID'].notna()]

# If you want to modify the original DataFrame in place
df.dropna(subset=['CustomerID'], inplace=True)

# Display info about removed rows
print(f"Original dataset shape: {df.shape}")
print(f"Cleaned dataset shape: {df_cleaned.shape}")
print(f"Rows removed: {df.shape[0] - df_cleaned.shape[0]}")
```
